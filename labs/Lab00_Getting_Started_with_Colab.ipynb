````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
plt.rcParams["figure.figsize"] = (8, 4)
# ---- Answer checks (do not edit) ----
import hashlib as _hl
import numpy as _np, pandas as _pd
def _norm(v, d=2):
    if isinstance(v, (_pd.Series, _pd.Index)): v = v.tolist()
    if isinstance(v, _np.ndarray): v = v.tolist()
    if isinstance(v, (list, tuple, set)):
        items = [_norm(x, d) for x in v]
        if isinstance(v, set): items = sorted(items)
        return "[" + "|".join(items) + "]"
    if isinstance(v, (bool, _np.bool_)): return str(bool(v))
    if isinstance(v, (int, float, _np.integer, _np.floating)):
        x = round(float(v), d)
        return f"{(0.0 if x == 0 else x):.{d}f}"
    return str(v).strip().lower()
def check_value(label, value, key, d=2, as_set=False):
    """Compares your answer with the expected one without revealing it."""
    try:
        if callable(value): value = value()
        if value is None or value is Ellipsis: raise ValueError
        if as_set: value = set(value)
        ok = _hl.sha256(_norm(value, d).encode()).hexdigest()[:12] == key
    except Exception:
        ok = False
    print(("✅ " if ok else "❌ ") + label)
def check(label, test, hint=""):
    try: ok = bool(test())
    except Exception: ok = False
    print(("✅ " if ok else "❌ ") + label + ("" if ok or not hint else "   Hint: " + hint))
print("Checks loaded.")
````
